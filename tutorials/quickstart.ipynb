# Semantic JSON 빠른 시작 / Quick Start

여신심사 문서를 SemanticDocument로 변환하고 의미 기반으로 검색합니다.

Compile credit-review text into a SemanticDocument and retrieve it semantically.

```python
from semantic_json import compile, SemanticRepository

# 한국어 문서를 의미 구조로 변환 (Compile a Korean document into semantic structure)
text = """B기업은 현재 원리금을 정상적으로 상환하고 있다. 다만 중장기 상환능력이 유지된다고 단정하기는 어렵다."""
doc = compile(text, document_id="sample_ko")
print(doc.to_json())
```

```python
# 의미 기반 저장소 구성 (Build a semantic repository)
repo = SemanticRepository()
repo.add(doc)
results = repo.search("medium-term repayment uncertainty", top_k=5)
print(repo.build_context(results))
```
